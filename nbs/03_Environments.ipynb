# Environments {#sec-environments}

In this section, we introduce three environment classes that are widely used in various disciplines, such as cognitive psychology or sustainability studies: a two-armed bandit task ( @sec-twoarmedbandit ), a social dilemma ( @sec-socialdilemma ), and a risk-reward dilemma ( @sec-rrdilemma ).

We will use these environments to compute our results in @sec-results .
