# Agents {#sec-agents}

In this section, we introduce various agent classes as well as biasing methods for the implementation of heuristics. @sec-ddrlbase contains the base functions for computing reward and value distributions. In @sec-ddrlsarsa , we introduce the functions that are specific to the SARSA algorithm of RL. By contrast, in @sec-ddrlac , we introduce those that are specific to the Actor-Critic version. Finally, @sec-biasing-methods describes three different ways of implementing the optimism/pessimism heuristic: the quantile method, the expectile method, and the weights method.
