```python
# Define environment
env = RiskRewardDilemma(pc=0.1, pr=0.2, rc=0.5, rr=1, rd=0)
```

```python
aei_unb = DDRLSarsa(env, nr_reward_bins=29, discretization_sigma=0.05, 
                    learning_rates=0.1, discount_factors=0.4, 
                    choice_intensities=50, Rmin=0, Rmax=1,
                    method="mean", use_prefactor=True)

aei_pes = DDRLSarsa(env, nr_reward_bins=29, discretization_sigma=0.05, 
                    learning_rates=0.1, discount_factors=0.4, 
                    choice_intensities=50, Rmin=0, Rmax=1,
                    method="weights", wU=0.1, use_prefactor=True)
```

```python
X = np.ones((1, 2, 2)) * .5
```

```python
aei_unb.Qisa(X)
```

```python
aei_pes.Qisa(X)
```

```python
gps = aei_unb.gridpoints
wdth = aei_unb.gridwidth
Qisa1 = aei_unb.Qisa(X)[0, 0, 0]
Qisa2 = aei_unb.Qisa(X)[0, 0, 1]
plt.bar(gps, aei_unb.Qisaq(X)[0,0,0,:], width=.9*wdth, color='blue', alpha=.4)
plt.bar(gps, aei_unb.Qisaq(X)[0,0,1,:], width=.9*wdth, color='red', alpha=.4)
plt.plot([Qisa1, Qisa1], [0, 0.01], color="blue", lw=5)
plt.plot([Qisa2, Qisa2], [0, 0.01], color="red", lw=5)
```

```python
from Methods import weights

Qisaq = aei_pes.Qisaq(X)  # Q-value distrib
Qisa = jnp.einsum('ijkl,l->ijk', Qisaq, gps)
i = 0; a = 1; s = 2
Qis = jnp.einsum(Qisa, [i, s, a], X, [i, s, a], [i, s])
# Baseline: mean of Qisaq
Qis_expanded = jnp.expand_dims(Qis, axis=2)  # shape (N, Z, 1)
baseline = jnp.repeat(Qis_expanded, aei_pes.M, axis=2)  # shape (N, Z, M)

w = jnp.ones((aei_pes.N, aei_pes.Z, aei_pes.M, len(gps)))
n = jnp.newaxis
upper = gps[n,n,n,:] >= baseline[:,:,:,n]
upper = upper * (aei_pes.wU - 1)
w = w + upper
wQisaq = weights(w, Qisaq, gps)
Qisa = jnp.einsum('ijkl,l->ijk', wQisaq, gps)  # take the new distrib's mean
```

```python
gps = aei_pes.gridpoints
wdth = aei_pes.gridwidth
Qisa1 = Qisa[0, 0, 0]
Qisa2 = Qisa[0, 0, 1]
plt.bar(gps, wQisaq[0,0,0,:], width=.9*wdth, color='blue', alpha=.4)
plt.bar(gps, wQisaq[0,0,1,:], width=.9*wdth, color='red', alpha=.4)
plt.plot([Qisa1, Qisa1], [0, 0.01], color="blue", lw=5)
plt.plot([Qisa2, Qisa2], [0, 0.01], color="red", lw=5)
```
