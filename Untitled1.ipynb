```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def softmax(beta, deltaQ):
    return 1 / (1 + np.exp(-beta*deltaQ))
```

```python
deltaQ = np.linspace(-2, 2, 50)
```

```python
betas = [0.5, 1, 4, 20, 50]
curves = np.zeros((len(betas), len(deltaQ)))
for i, beta in enumerate(betas):
    curves[i, :] = softmax(beta, deltaQ)
```

```python
for i, beta in enumerate(betas):
    plt.plot(deltaQ, curves[i, :], label=str(beta))
    plt.legend()
```
